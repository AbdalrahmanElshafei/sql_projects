```python
import pandas as pd
import pyodbc 

from sqlalchemy import create_engine
from urllib.parse import quote_plus
```

```python
connection_string = quote_plus(
    "DRIVER={ODBC Driver 18 for SQL Server};"
    "SERVER=localhost;"
    "DATABASE=Ecommerce_Orders;"
    "Trusted_Connection=yes;"
    "TrustServerCertificate=yes;"
)

engine = create_engine(
    f"mssql+pyodbc:///?odbc_connect={connection_string}"
)
```

```python
customers = pd.read_parquet(
    r"C:\Users\ELmohannd\Desktop\New folder\customers.parquet"
)

products = pd.read_parquet(
    r"C:\Users\ELmohannd\Desktop\New folder\products.parquet"
)

orders = pd.read_parquet(
    r"C:\Users\ELmohannd\Desktop\New folder\orders.parquet"
)

order_items = pd.read_parquet(
    r"C:\Users\ELmohannd\Desktop\New folder\order_items.parquet"
)

payments = pd.read_parquet(
    r"C:\Users\ELmohannd\Desktop\New folder\payments.parquet"
)

returns = pd.read_parquet(
    r"C:\Users\ELmohannd\Desktop\New folder\returns.parquet"
)
```

```python
print("customers:", customers.shape)
print("products:", products.shape)
print("orders:", orders.shape)
print("order_items:", order_items.shape)
print("payments:", payments.shape)
print("returns:", returns.shape)
```

```python
for name, df in {
    "categories": categories,
    "customers": customers,
    "products": products,
    "orders": orders,
    "order_items": order_items,
    "payments": payments,
    "returns": returns
}.items():
    print(f"\n--- {name} ---")
    print(df.dtypes)
```

```python
tables = {
    "categories": categories,
    "customers": customers,
    "products": products,
    "orders": orders,
    "order_items": order_items,
    "payments": payments,
    "returns": returns
}

for table_name, df in tables.items():
    print(f"Loading {table_name}...")

    df.to_sql(
        table_name,
        con=engine,
        if_exists="append",
        index=False,
        chunksize=5000
    )

    print(f"{table_name} loaded successfully!")
```
